# Importing Libraries


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
```

```python
from sklearn import datasets
from sklearn.linear_model import LogisticRegression
```

# load the data


```python
iris = datasets.load_iris()
```

```python
iris.keys()
```

```python
X = iris["data"][:, 3:]
```

```python
y = (iris["target"] == 2).astype(int)
```

```python
X_s = iris["data"][:, (2, 3)]
```

```python
y_s = iris["target"]
```

# Train a model


## baseline


```python
log_reg = LogisticRegression()
```

```python
log_reg.fit(X, y)
```

## with softmax


```python
softmax_reg = LogisticRegression(solver="lbfgs", C=10)
```

```python
softmax_reg.fit(X_s, y_s)
```

# Predict


## with logistic reg


```python
X_new = np.linspace(0, 3, 1000).reshape(-1, 1)
y_proba = log_reg.predict_proba(X_new)
```

```python
plt.plot(X_new, y_proba[:, 1], "g-", label="Iris-virginica")
plt.plot(X_new, y_proba[:, 0], "b--", label="Not Iris-Virginica")
```

```python
log_reg.predict([[1.7], [1.5]])
```

## softmax


```python
softmax_reg.predict([[5, 2]])
```

```python
softmax_reg.predict_proba([[5, 2]])
```

```python
x1, x2 = np.meshgrid(
    np.linspace(0, 7, 500), np.linspace(0, 3.5, 300)  # petal length  # petal width
)

grid_points = np.c_[x1.ravel(), x2.ravel()]
```

```python
predicted_classes = softmax_reg.predict(grid_points)

# Probability of class 2: Iris-Virginica
virginica_probability = softmax_reg.predict_proba(grid_points)[:, 2]

predicted_classes = predicted_classes.reshape(x1.shape)
virginica_probability = virginica_probability.reshape(x1.shape)
```

```python
background_colors = ListedColormap(
    [
        "#fff9a8",  # Setosa region
        "#9999ee",  # Versicolor region
        "#99ee99",  # Virginica region
    ]
)

plt.figure(figsize=(10, 6))

plt.contourf(x1, x2, predicted_classes, cmap=background_colors, alpha=0.8)

markers = ["o", "s", "^"]
labels = ["Iris-Setosa", "Iris-Versicolor", "Iris-Virginica"]

for class_number in range(3):
    plt.scatter(
        X_s[y_s == class_number, 0],
        X_s[y_s == class_number, 1],
        marker=markers[class_number],
        label=labels[class_number],
        edgecolor="black",
        s=45,
    )
plt.xlabel("Petal length (cm)")
plt.ylabel("Petal width (cm)")
plt.title("Softmax Regression Decision Boundaries")

plt.xlim(0, 7)
plt.ylim(0, 3.5)

plt.legend()
plt.grid(alpha=0.2)
plt.show()
```
